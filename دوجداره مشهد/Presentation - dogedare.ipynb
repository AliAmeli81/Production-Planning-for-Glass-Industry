# spaser

```python
model1 = joblib.load("xgb_model.pkl")

print("✅ مدل لود شد.")
print("نوع:", type(model1).__name__)
print("Feature names:", list(model1.feature_names_in_))
# انتظار: ['طول', 'عرض', 'ضخامت']
```

```python
# ========================== Feature Names ==========================
FEATURE_NAMES = list(model1.feature_names_in_)


# ========================== Predict Function ==========================
def predict_xgb(length, width, thickness):
    """
    Inputs:
        length    : طول
        width     : عرض
        thickness : ضخامت
    
    Output:
        predicted time (float)
    """
    # نگاشت هوشمند مقادیر به اسم‌های واقعی مدل
    values = {}
    for name in FEATURE_NAMES:
        n = str(name)
        if "طول" in n or "length" in n:
            values[name] = length
        elif "عرض" in n or "width" in n:
            values[name] = width
        elif "ضخامت" in n or "thick" in n:
            values[name] = thickness
        else:
            raise ValueError(f"ستون ناشناخته: {n}")

    # DataFrame با ترتیب دقیق مدل
    features = pd.DataFrame(
        [[values[name] for name in FEATURE_NAMES]],
        columns=FEATURE_NAMES
    )

    # فقط predict — چون Regression هست
    prediction = model1.predict(features)[0]
    return prediction


# ========================== Example Usage ==========================
if __name__ == "__main__":
    result = predict_xgb(length=129, width=109, thickness=6)
    print(f"پیش‌بینی زمان: {result:.4f}")
```

# gas

```python
modelgass = joblib.load("gass.pkl")

print("✅ مدل لود شد.")
print("نوع:", type(modelgass).__name__)
print("کلاس‌ها:", list(modelgass.classes_))
print("Feature names:", list(modelgass.feature_names_in_))
```

```python
# ========================== Feature Names ==========================
FEATURE_NAMES5gass = list(modelgass.feature_names_in_)
# انتظار: ['length', 'width', 'thickness']


# ========================== Predict Function ==========================
def predict_koreh(length, width, thickness, top_n=5):
    """
    Inputs:
        length    : طول
        width     : عرض
        thickness : ضخامت
        top_n     : چند کلاس برتر نمایش داده بشه
    
    Output:
        prediction : کلاس پیش‌بینی‌شده (تایم)
        prob_df    : دیتافریم احتمالات هر کلاس (مرتب‌شده نزولی)
    """
    # دیکشنری مقادیر با نگاشت هوشمند
    values = {}
    for name in FEATURE_NAMES5gass:
        n = str(name)
        if "length" in n or "طول" in n:
            values[name] = length
        elif "width" in n or "عرض" in n:
            values[name] = width
        elif "thick" in n or "ضخامت" in n:
            values[name] = thickness
        else:
            raise ValueError(f"ستون ناشناخته: {n}")

    # DataFrame با ترتیب دقیق مدل
    features = pd.DataFrame(
        [[values[name] for name in FEATURE_NAMES5gass]],
        columns=FEATURE_NAMES5gass
    )

    # پیش‌بینی
    prediction = modelgass.predict(features)[0]
    probs = modelgass.predict_proba(features)[0]

    # دیتافریم احتمالات
    prob_df = (
        pd.DataFrame({
            "time_class": modelgass.classes_,
            "probability": probs
        })
        .sort_values("probability", ascending=False)
        .head(top_n)
        .reset_index(drop=True)
    )

    return prediction, prob_df


# ========================== Example Usage ==========================
if __name__ == "__main__":
    length    = 500
    width     = 500
    thickness = 10

    pred, prob_df = predict_koreh(length, width, thickness)

    print(f"پیش‌بینی کلاس: {pred}")
    print("\nاحتمالات:")
    print(prob_df.to_string(index=False))
```
